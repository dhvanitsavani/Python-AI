## Feature Extraction Project

### Task

#### Read anime.csv file using pandas & perform below operations on it.
* make a new column for episode count
* make a new column for time stamp
* which anime has the highest score
* give me top 5 highest scoring anime
* which anime has the highest episode count
* animes with top 5 episode count
* which is the longest running anime

### Implementation

#### Read csv file & create a dataframe

```python
import numpy as np
import pandas as pd

anime_df = pd.read_csv('anime.csv')
anime_df
```

#### Make column for episodes count

```python
def extract_episodes_count(title):
    is_eps_num = False
    ans = ""

    for i in title:
        if i == ' ' and is_eps_num:
            return int(ans)
        if is_eps_num:
            ans += i
        if i == '(':
            is_eps_num = True
```

```python
anime_df['Episodes'] = anime_df['Title'].apply(extract_episodes_count)
anime_df.head()               # Only shows first five rows
```

#### Make column for time stamp

```python
import re

time_stamp_pattern = r'[A-Za-z]{3} \d{4} - [A-Za-z]{3} \d{4}'

def extract_time_stamp(title):
    match = re.search(time_stamp_pattern, title)

    if match:
        return match.group()
```

```python
anime_df['Time Stamp'] = anime_df['Title'].apply(extract_time_stamp)
anime_df.head()
```

#### Make column for total months of time stamp

```python
months_codes = {
    'jan': 1, 'feb': 2, 'mar': 3,
    'apr': 4, 'may': 5, 'jun': 6,
    'jul': 7, 'aug': 8, 'sep': 9,
    'oct': 10, 'nov': 11, 'dec': 12,
}

month_pattern = r'[a-zA-Z]{3}'
year_pattern = r'\d{4}'
```

```python
def extract_total_months(title):
    months = re.findall(month_pattern, title)
    month_1 = months[0].lower()
    month_2 = months[1].lower()
    month_sub = months_codes[month_2] - months_codes[month_1]

    years = re.findall(year_pattern, title)
    year_1 = int(years[0])
    year_2 = int(years[1])
    year_sub = (year_2 - year_1) * 12

    return month_sub + year_sub
```

```python
anime_df['Total Months'] = anime_df['Time Stamp'].apply(extract_total_months)
anime_df.head()
```

#### Highest score

```python
anime_df[anime_df['Rank'] == 1]
```

#### Top 5 highest score anime

```python
anime_df[anime_df['Rank'] <= 5]
```

#### Highest episode count

```python
anime_df[anime_df['Episodes'] == anime_df['Episodes'].max()]
```

#### Top 5 highest episodes count anime

```python
highest_episodes_top_5_df = pd.DataFrame(columns=anime_df.columns)
dummy_anime_df = anime_df.copy()

for i in range(5):
    highest_episodes_top_5_df = pd.concat([highest_episodes_top_5_df, dummy_anime_df.loc[[dummy_anime_df['Episodes'].idxmax()]]])
    dummy_anime_df.drop(dummy_anime_df['Episodes'].idxmax(), inplace=True)

del dummy_anime_df
highest_episodes_top_5_df
```
