### Task

#### Read anime.csv file using pandas & perform below operations on it.
- Which country has the highest population
- What is the capital of the country with highest population
- Which country has the least population
- Give me top 5 countries with highest democratic score
- How many total regions are there
- How many countries lie in Eastern Europe region
- Who is the political leader of the 2nd highest populated country

### Implementation

#### Read csv file & create a datagrame

```python
import pandas as pd

countries_df = pd.read_csv('Countries.csv')
countries_df.head()
```

#### Highest population country

```python
countries_df[countries_df['population'] == countries_df['population'].max()]
```

#### Capital of highest populated country

```python
countries_df[countries_df['population'] == countries_df['population'].max()]['capital_city']
```

#### Least population country

```python
countries_df[countries_df['population'] == countries_df['population'].min()]
```

#### Top 5 highest democratic score countries

```python
dummy_countries_df = countries_df.copy()
top_5_democracy_score_countries_df = pd.DataFrame(columns=countries_df.columns)

for i in range(5):
    top_5_democracy_score_countries_df = pd.concat([top_5_democracy_score_countries_df, dummy_countries_df.loc[[dummy_countries_df['democracy_score'].idxmax()]]])
    dummy_countries_df.drop(dummy_countries_df['democracy_score'].idxmax(), inplace=True)

del dummy_countries_df

top_5_democracy_score_countries_df
```

#### Total regions

```python
countries_df['region'].nunique()
```

#### Total countries in Eastern Europe region

```python
int(countries_df[countries_df['region'] == 'Eastern Europe']['country'].count())
```

```python
dummy_countries_df = countries_df.copy()
dummy_countries_df.drop(dummy_countries_df['population'].idxmax(), inplace=True)

print(dummy_countries_df[dummy_countries_df['population'] == dummy_countries_df['population'].max()]['political_leader'].iloc[0])
del dummy_countries_df
```
