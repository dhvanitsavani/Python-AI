### Array Indexing

#### Vector

```python
arr = np.arange(11,21)
print(arr)
print()

print(arr[5])      # Single element
print(arr[2:7:2])  # Range
```

#### Matrix

```python
arr = np.arange(1,31).reshape(6,5)
print(arr)
print()

print('Row : ', arr[5])           # Row
print('Single element : ', arr[5,2])         # Single element
print('Range : ', arr[1:3,:4], sep='\n')      # Range
print('Column : ', arr[:,3])         # Column
```

#### Deep copy, Shallow copy, & Assignment

```python
a = np.arange(1,11)
```

##### Deep copy

```python
deep_copy = a.copy()
deep_copy += 10

print(deep_copy)
print(a)
```

##### Shallow copy

```python
shallow_copy = a.view()
shallow_copy += 10

print(shallow_copy)
print(a)

print(shallow_copy is a)
```

##### Assignment

```python
assigned_object = a
assigned_object += 100

print(assigned_object)
print(a)

print(assigned_object is a)
```
