# Numpy

## Advanced

```python
import numpy as np
```

### Triogonometric Functions

#### Basic Functions

```python
angle = np.pi / 2
print(np.sin(angle))             # Angle in radian
print(np.cos(angle))
print(np.tan(angle))
```

#### For Arrays

```python
angles = np.array([0, np.pi / 2, np.pi])

print(np.sin(angles))
```

#### Conversion between degrees & radians

```python
print(np.deg2rad(90))
print(np.rad2deg(np.pi))
```

#### Inverse Functions

```python
inverse_angle = 1

print(np.arcsin(inverse_angle))
print(np.arccos(inverse_angle))
print(np.arctan(inverse_angle))
```

#### Hyperbolic Functions

```python
print(np.sinh(angle))
print(np.cosh(angle))
print(np.tanh(angle))

print(np.arcsinh(inverse_angle))
print(np.arccosh(inverse_angle))
print(np.arctanh(inverse_angle - 0.1))
```

### Useful Functions

#### exp

```python
print(np.exp(0))              # Returns exponential value
print(np.exp(1))
print(np.exp(2))

print(np.expm1(0))            # Return e**x - 1
print(np.expm1(1))
print(np.expm1(2))
```

```python
arr = np.array([1,2,3,4,5])
print(np.exp(arr))
print(np.expm1(arr))
```

#### log

```python
print(np.log(arr))
print(np.log10(arr))
print(np.log2(arr))
```

#### fromfunction

##### Using def

```python
def sum(x, y):
    return x + y

arr = np.fromfunction(sum, (3, 5), dtype=int)
arr
```

##### Using lambda

```python
arr = np.fromfunction(lambda x, y: (x+1) * (y+1), (4, 4), dtype=int)
arr
```

#### flat

```python
arr = np.arange(1,11)
for i in arr.flat:               # This lets you to work with each elements indivisually of numpy array.
    print(i)
```

#### ellipsis (...)

##### 3D array

```python
arr_3D = np.arange(2*3*4).reshape(2,3,4)

print(arr_3D[:, :, 2])
print()
print(arr_3D[..., 2])
```

##### 4D array

```python
arr_4D = np.arange(2*3*4*5).reshape(2,3,4,5)

print(arr_4D[0, :, :, 2])
print()
print(arr_4D[0, ..., 2])
```
