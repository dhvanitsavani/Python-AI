### Triogonometric Functions

```python
inverse_angle = 1

print(np.arcsin(inverse_angle))
print(np.arccos(inverse_angle))
print(np.arctan(inverse_angle))
```

#### fromfunction

This numpy array creation method applies the given function to all the elements while creating the array.

##### Using def

```python
def sum(x, y):
    return x + y

arr = np.fromfunction(sum, (3, 5), dtype=int)
arr
```

##### Using lambda

```python
arr = np.fromfunction(lambda x, y: (x+1) * (y+1), (4, 4), dtype=int)
arr
```

### Dimensional Indexing Tools

#### ellipsis (...)

##### 3D array

```python
arr_3D = np.arange(2*3*4).reshape(2,3,4)

print(arr_3D[:, :, 2])
print()
print(arr_3D[..., 2])
```

##### 4D array

```python
arr_4D = np.arange(2*3*4*5).reshape(2,3,4,5)

print(arr_4D[0, :, :, 2])
print()
print(arr_4D[0, ..., 2])
```

#### newaxis

##### basics

```python
# newaxis adds dimension to the resulting selection
arr = np.arange(1,5)
print(arr, arr.shape, sep='\n')
print()

new_arr_1 = arr[:, np.newaxis]
print(new_arr_1, new_arr_1.shape, sep='\n')
print()

new_arr_2 = arr[np.newaxis, :]
print(new_arr_2, new_arr_2.shape, sep='\n')
```

##### newaxis for broadcasting

```python
a = np.array([1, 2, 3])
b = np.array([10, 20])

# a + b will produce an error.
# but we can use newaxis to do broadcasting in these two arrays.

a[:, np.newaxis] + b
```

### Advanced Indexing

#### Basics

##### 1D Arrays

```python
arr = np.array([10,20,30,40,50])

# Advanced Integer Indexing
indices = np.array([0, 2])
print('Advanced Integer Indexing : ', arr[indices])

# Using python list
print('Using python list : ', arr[[0, 2]])

# Repeat indices
print('Repeated indices : ', arr[[0, 2, 2, 3, 0]])

# Negative indices
print('Negative indices : ', arr[[0, -3, 1]])
```

##### 2D Arrays

```python
arr_2D = np.arange(9).reshape(3,3)

# Selecting rows
print(arr_2D[[0, 2]])
print()

# Spesific element
# Numpy will pair the indices
row = [0]
col = [2]
print(arr_2D[row, col])

rows = [0, 1, 2]
cols = [0, 1, 2]
print(arr_2D[rows, cols])
```

##### copy

```python
# Advanced Indexing creates a copy
arr_temp = np.array([10, 20, 30, 40])
x = arr_temp[[0, 2]]
x[0] = 999
print(x, arr_temp, sep='\n')
```

#### np.ix_()

```python
# In indexing spesific elements from 2D array, advanced indexing pairs the corresponding indices. But np.ix_() pairs all the indices and make all the combinations.

print('Advanced Indexing : ', arr_2D[rows, cols])
print()
print('np.ix_() : ', arr_2D[np.ix_(rows, cols)], sep='\n')
```

#### Broadcasting with Advaned Indexing

```python
# When advanced index arrays have the same shape, advanced index arrays have the same shape
# Broadcasting comes in when index arrays have different shapes.

rows = np.array([               # Shape : (2,1)
    [0],
    [1]
])
cols = np.array([0, 2, 1])      # Shape : (3,) -> (1,3) because Numpy treats cols's shape as (1, 3)
print(arr_2D[rows, cols])

# Broadcasting
#
# - dimensions :
#   2 1
#   1 3
#
# - Compare from the right :
#   1 and 3 -> compatible
#   2 and 1 -> compatible
#   Because one of the dimensions is 1, Numpy can expand it.
#   Both becomes (2,3) because Numpy choose max of dimensions - max of right side is 3 and max of left side is 2 so shape becomes (2,3)
#
# - Expand rows
#   [[0],  ->  [[0, 0, 0],
#    [1]        [1, 1, 1]]
#
# - Expand cols
#   [0, 2, 1]  ->  [[0, 2, 1],
#                   [0, 2, 1]]
#
# - Pair the coordinates and get the values.
```

### Field Access

#### Basics

```python
arr = np.array([
    ("Dhvanit", 19, 93),
    ("Het", 18, 81),
    ("Sujal", 21, 60)
], dtype=[
    ("name", "U20"),
    ("age", "i4"),
    ("marks", "i4")
])

print(arr)
print(arr["name"])

print(arr['age'][1])
print(arr[['name', 'marks']])

print(arr.dtype.names)
```

#### View

```python
# Field access creates a view.
marks = arr['marks']
marks[1] = 100

print(marks, arr, sep='\n')
```

#### Dynamic indices

##### First element

```python
arr = np.arange(20).reshape(4,5)

index = (0,) * arr.ndim

print(arr[index])
```

##### Last dimension

```python
arr = np.arange(120).reshape(2,3,4,-1)

index = (slice(None),) * (arr.ndim - 1) + (-1,)

print(arr[index])       # equivalent to arr[:,:,:,-1] or arr[...,-1]
```

#### Function example

```python
def set_first_element(arr, el):
    index = (0,) * arr.ndim
    arr[index] = el

def access_first_element(arr):
    index = (0,) * arr.ndim
    return arr[index]
```

```python
arr1 = np.arange(5)
arr2 = np.arange(12).reshape(3,4)
arr3 = np.arange(30).reshape(2,3,-1)
arr4 = np.arange(120).reshape(2,3,4,-1)
arr5 = np.arange(300).reshape(1,2,3,5,-1)

set_first_element(arr1, 10)
set_first_element(arr2, 5)
set_first_element(arr3, 17)
set_first_element(arr4, -5)
set_first_element(arr5, 0)
```

```python
print(access_first_element(arr1), access_first_element(arr2), access_first_element(arr3), access_first_element(arr4), access_first_element(arr5), sep='\n')
```
