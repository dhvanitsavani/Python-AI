# Numpy

## Advanced

```python
import numpy as np
```

### Triogonometric Functions

#### Basic Functions

```python
angle = np.pi / 2
print(np.sin(angle))             # Angle in radian
print(np.cos(angle))
print(np.tan(angle))
```

#### For Arrays

```python
angles = np.array([0, np.pi / 2, np.pi])

print(np.sin(angles))
```

#### Conversion between degrees & radians

```python
print(np.deg2rad(90))
print(np.rad2deg(np.pi))
```

#### Inverse Functions

```python
inverse_angle = 1

print(np.arcsin(inverse_angle))
print(np.arccos(inverse_angle))
print(np.arctan(inverse_angle))
```

#### Hyperbolic Functions

```python
print(np.sinh(angle))
print(np.cosh(angle))
print(np.tanh(angle))

print(np.arcsinh(inverse_angle))
print(np.arccosh(inverse_angle))
print(np.arctanh(inverse_angle - 0.1))
```

### Useful Functions

#### exp

```python
print(np.exp(0))              # Returns exponential value
print(np.exp(1))
print(np.exp(2))

print(np.expm1(0))            # Return e**x - 1
print(np.expm1(1))
print(np.expm1(2))
```

```python
arr = np.array([1,2,3,4,5])
print(np.exp(arr))
print(np.expm1(arr))
```

#### log

```python
print(np.log(arr))
print(np.log10(arr))
print(np.log2(arr))
```

#### fromfunction

This numpy array creation method applies the given function to all the elements while creating the array.

##### Using def

```python
def sum(x, y):
    return x + y

arr = np.fromfunction(sum, (3, 5), dtype=int)
arr
```

##### Using lambda

```python
arr = np.fromfunction(lambda x, y: (x+1) * (y+1), (4, 4), dtype=int)
arr
```

#### flat

```python
arr = np.arange(1,11)
for i in arr.flat:               # This lets you to work with each elements indivisually of numpy array.
    print(i)
```

### ellipsis (...)

#### 3D array

```python
arr_3D = np.arange(2*3*4).reshape(2,3,4)

print(arr_3D[:, :, 2])
print()
print(arr_3D[..., 2])
```

#### 4D array

```python
arr_4D = np.arange(2*3*4*5).reshape(2,3,4,5)

print(arr_4D[0, :, :, 2])
print()
print(arr_4D[0, ..., 2])
```

### I/O in Numpy

#### Reading from files

##### Creating array using file data

```python
# genfromtxt will convert file's data to numpy array.
arr = np.genfromtxt("files/numbers_1.txt")
arr
```

```python
# delimeter defines which seperator is used in file.
arr = np.genfromtxt("files/numbers_2.csv", delimiter=",")
arr
```

##### Datatype

```python
# dtype converts datatype to given datatype.
# Default datatype is float
arr = np.genfromtxt("files/numbers_1.txt", delimiter=",", dtype=int)
print(arr)
arr = np.genfromtxt("files/numbers_1.txt", delimiter=",", dtype=str)
print(arr)
```

##### autostrip argument

```python
# autostrip automatically strips the string with whitespaces.
arr = np.genfromtxt("files/students_marks_1.csv", delimiter=",", dtype=None)
print(arr)
arr = np.genfromtxt("files/students_marks_1.csv", delimiter=",", dtype=None, autostrip=True)
print(arr)
```

#### names argument

##### Using names from file

```python
# names argument gives names to the columns.
arr = np.genfromtxt("files/students_data_1.csv", delimiter=",", dtype=None, autostrip=True, names=True)
print(arr)
print()

print(arr.dtype)
print()

print(arr['Roll'])
print(arr['Name'])
```

##### Using names from variable

```python
names = "Name, Marks"
arr = np.genfromtxt("files/students_marks_1.csv", delimiter=",", dtype=None, autostrip=True, names=names)
arr
```

#### Skipping headers and footers

##### Skipping lines from starting of the file.

```python
# arr = np.genfromtxt("files/numbers_3.txt")
# print(arr)

# Above code will produce an error.

arr = np.genfromtxt("files/numbers_3.txt", skip_header=1)
print(arr)
```

##### Skipping lines from ending of the file.

```python
# arr = np.genfromtxt("files/numbers_4.txt")
# print(arr)

# Above code will produce an error.

arr = np.genfromtxt("files/numbers_4.txt", skip_footer=1)
print(arr)
```

#### Other useful arguments

##### comments argument

Defines which marker is for comments in file.

Default value of it is #.

```python
arr = np.genfromtxt("files/numbers_5.csv", delimiter=",", comments="//")
arr
```

##### usecols argument

Defines which columns should be existed in array.

```python
arr = np.genfromtxt("files/numbers_1.txt", usecols=[0, 2])
arr
```

##### unpack argument

Distributes the columns to variables.

```python
a, b, c = np.genfromtxt("files/numbers_1.txt", unpack=True)
print(a, b, c, sep='\n')
```

##### converters argument

converts perform given operations while creating array.

```python
arr = np.genfromtxt("files/students_marks_1.csv", delimiter=",", autostrip=True, dtype=None, converters={0: lambda x: x.upper()})
arr
```

Use cases of converters : to convert text case, to replace missing values, to create date object for dates, etc.

##### defaultfmt arguments

```python
# By default, names are given as f0, f1, f2,.....
arr = np.genfromtxt("files/numbers_1.txt", dtype=[int, int, int])
print(arr.dtype)

# We can define to give names by using defaultfmt argument.
arr = np.genfromtxt("files/numbers_1.txt", dtype=[int, int, int], defaultfmt="column_%d")
print(arr.dtype)
```

#### Missing and Filling values

##### Missing values

Whenever an value is empty is file, it will be considered as a missing value in array.

```python
arr = np.genfromtxt("files/numbers_6.csv", delimiter=",")
print(arr)

#  float     nan
#  int       -1
#  str       ""
#  boolean   False
```

##### Filling Values argument

filling_values argument fills the missing values as given.

```python
arr = np.genfromtxt("files/numbers_6.csv", delimiter=",", filling_values=-1)
print(arr)
```

#### Reading from URL or file object

##### URL

```python
with open("files/numbers_1.txt") as f:
    arr = np.genfromtxt(f)

arr
```

##### File Object

```python
from io import StringIO

data = StringIO("""1,2,3
    4,5,6
    7,8,9
""")

arr = np.genfromtxt(data, delimiter=",")
arr
```

#### Validating names

##### replace_space

replace_space argument replaces the spaces in names of columns with given value.

```python
arr = np.genfromtxt("files/students_marks_2.csv", delimiter=",", dtype=['U8', int], names=True, replace_space="_", skip_footer=4)
arr
```

##### deletechars

deletechars argument deletes the given characters existing in names of columns.

```python
arr = np.genfromtxt("files/students_marks_2.csv", delimiter=",", dtype=['U8', int], names=True, deletechars="-%", skip_header=4)
arr
```

##### case_sensetive

case_sensetive defines what the case should be of columns names of array.

```python
arr = np.genfromtxt("files/students_data_1.csv", delimiter=",", dtype=None, autostrip=True, names=True, case_sensitive=False)
arr.dtype

# case_sensetive=False or case_sensetive='upper' converts case to upper.
# case_sensetive='lower' converts case to lower.
# Default value of case_sensetive is True which remains the case as it is.
```

##### exclude_list

exclude_list don't allow the given values as names of columns of array.

```python
arr = np.genfromtxt("files/students_data_2.csv", delimiter=",", autostrip=True, dtype=None, names=True, excludelist=["Class"])
arr
```
