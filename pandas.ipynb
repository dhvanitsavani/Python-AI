### Dataframes

#### Creation

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
df1 = pd.DataFrame(data)
df1
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
df2 = pd.DataFrame(data_list)
df2
```

```python
columns = ["Name","Age","City","Salary"]
df2 = pd.DataFrame(data_list,columns =columns)
df2
```

#### Selecting columns

```python
print(df2["Name"])
df2[["Name", "City"]]
```

#### Creating columns

```python
df2["Designation"] = ["Doctor","Eng.","Doctor","Eng."]
df2
```

#### Removing columns

```python
df2.drop("Designation", axis=1, inplace=True) # If inplace is not set to True, the deletation will only be for current code not for memory.
df2
```

#### Removing rows

```python
df2.drop(0, axis=0)
```

```python
df2
```

#### Selecting rows

```python
df2.loc[0]
```

```python
df2.loc[[0,2]]
```

```python
df2.iloc[1]
```

#### Selecting subsets of rows and columns

```python
df2.loc[[0,1]][['Name', 'City']]
```

#### Conditional selection

```python
df2[df2['Age'] > 30]
```

```python
df2[(df2["Age"] > 30) & (df2["City"] == "Paris")]
```
